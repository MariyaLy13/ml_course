1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
import numpy as np
import matplotlib.pyplot as plt

# Визначення функції та її похідної
x = np.linspace(0, 10, 100)
f = 10 * x - 5
df = np.full_like(x, 10)  # Похідна f'(x) = 10

# Створення графіку
plt.figure(figsize=(10, 5))
plt.plot(x, f, label='f(x) = 10x - 5', color='blue')
plt.plot(x, df, label="f'(x) = 10", color='green')

# Позначення точки та дотичної лінії
x_point = 5
f_point = 10 * x_point - 5
tangent_line = 10 * (x - x_point) + f_point  # y - y1 = m(x - x1)

plt.scatter(x_point, f_point, color='red')  # точка на кривій
plt.plot(x, tangent_line, '--', color='red', label='Tangent at x = 5')

# Додавання легенди, міток та заголовка
plt.title("Графік функції та її похідної")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```

Аналіз:
1. Функція f(x)=10x-5
- Це лінійна залежність.
- Кожна додаткова година навчання підвищує оцінку на 10 балів.
- При нульовій кількості годин оцінка становить –5 (математично, хоча в реальності оцінка не може бути від’ємною).
2. Похідна f'(x)=10
- Похідна показує, як швидко змінюється оцінка при зміні кількості годин.
- Оскільки похідна стала, темп зростання постійний.
- Це означає, що кожна година навчання дає однаковий приріст результату.

2. Використовуючи Python і `SymPy`, обчисліть похідні для функцій $ f(x) = \sin(x) $ та $ f(x) = \ln(x) $. Зобразіть ці функції та їхні похідні за допомогою `matplotlib`.



```python
import sympy as sp

# Символьна змінна
x = sp.symbols('x')

# Функції
f_sin = sp.sin(x)
f_ln = sp.log(x)

# Похідні
derivative_f_sin = sp.diff(f_sin, x)
derivative_f_ln = sp.diff(f_ln, x)

print("Похідна sin(x):", derivative_f_sin)
print("Похідна ln(x):", derivative_f_ln)
```

```python
# Перетворення в числові функції
f_sin_num = sp.lambdify(x, f_sin, 'numpy')
df_sin_num = sp.lambdify(x, derivative_f_sin, 'numpy')

f_ln_num = sp.lambdify(x, f_ln, 'numpy')
df_ln_num = sp.lambdify(x, derivative_f_ln, 'numpy')

# Точки для графіків
x_vals = np.linspace(0.1, 10, 400)  # уникаємо x=0 для ln(x)

# Побудова графіків
plt.figure(figsize=(12, 5))

# sin(x) та його похідна
plt.subplot(1, 2, 1)
plt.plot(x_vals, f_sin_num(x_vals), label='sin(x)')
plt.plot(x_vals, df_sin_num(x_vals), label="sin'(x) = cos(x)")
plt.title("sin(x) та його похідна")
plt.legend()
plt.grid(True)

# ln(x) та його похідна
plt.subplot(1, 2, 2)
plt.plot(x_vals, f_ln_num(x_vals), label='ln(x)')
plt.plot(x_vals, df_ln_num(x_vals), label="(ln(x))' = 1/x")
plt.title("ln(x) та його похідна")
plt.legend()
plt.grid(True)

plt.show()
```

Aналіз:
1. sin(x) та cos(x)
- sin(x) коливається між –1 та 1.
- Його похідна cos(x) також коливається, але зсунута по фазі.
- Похідна показує швидкість зміни sin(x):
коли sin(x) зростає — cos(x) додатний, коли спадає — cos(x) від’ємний.
2. ln(x) та 1/x
- ln(x) зростає повільно, особливо після x > 3.
- Похідна 1/x показує, що швидкість зростання зменшується зі збільшенням x.
- При x → 0⁺ похідна різко зростає, бо ln(x) змінюється дуже швидко.


3. Загальні витрати компанії від обсягу виробництва задані функцією $ f(x) = x^2 - 5x + 1 $, де $ x $ — кількість вироблених одиниць. Обчисліть похідну цієї функції з допомогою SymPy та візуалізуйте значення функції і її похідної на відрізку від 0 до 100.

```python
# Символьна змінна
x = sp.symbols('x')

# Функція загальних витрат
f_exp = x**2 - 5*x + 1

# Похідна
derivative_f_exp = sp.diff(f_exp, x)

print("Похідна функції витрат:", derivative_f_exp)
```

```python
# Перетворення в числові функції
f_exp_num = sp.lambdify(x, f_exp, 'numpy')
df_exp_num = sp.lambdify(x, derivative_f_exp, 'numpy')

# Точки для графіка
x_vals = np.linspace(0, 100, 400)

# Побудова одного графіка
plt.figure(figsize=(10, 6))

plt.plot(x_vals, f_exp_num(x_vals), label='C(x) = x^2 - 5x + 1')
plt.plot(x_vals, df_exp_num(x_vals), label="C'(x) = 2x - 5", linewidth=2)

plt.title("Функція витрат та її похідна на одному графіку")
plt.xlabel("x")
plt.ylabel("Значення")
plt.grid(True)
plt.legend()

plt.show()
```

Aналіз
1. Функція витрат
- Це парабола, що спочатку спадає, а потім зростає.
- Мінімум досягається там, де похідна дорівнює нулю:
2x-5=0, x=2.5
Отже, при виробництві приблизно 2.5 одиниць витрати мінімальні.
2. Похідна
- Показує, як швидко змінюються витрати при збільшенні виробництва.
- Для x<2.5 похідна від’ємна → витрати зменшуються.
- Для x>2.5 похідна додатна → витрати зростають.
