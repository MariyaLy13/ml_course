Аналіз середнього, медіани, моди та розподілу показує, що найчастіше значення дорівнює 0. Пропущені значення є структурними: вони означають відсутність запитів, а не втрату даних. Тому оптимальним способом обробки пропусків є заміна NaN на 0. Це зберігає логіку даних і не спотворює розподіл.

Тому я вирішила заповнити дані модою для цих стовпців. Для цього я створю функцію impute_with_mode.

```python
def impute_with_mode(df, col):
    """
    Impute missing values in a column using its mode
    and create an indicator column showing which rows were imputed.
    """
    # Indicator column
    df[col + '_was_imputed'] = df[col].isna()
    
    # Mode value
    mode_value = df[col].mode()[0]
    
    # Impute
    df[col] = df[col].fillna(mode_value)
    
    return df
```

```python
cols = [
    'AMT_REQ_CREDIT_BUREAU_YEAR',
    'AMT_REQ_CREDIT_BUREAU_MON',
    'AMT_REQ_CREDIT_BUREAU_WEEK'
]

for col in cols:
    df = impute_with_mode(df, col)
```

# ДЗ 2. Аналіз і заповнення пустих значень у колонках з малим відсотком пропущених

За прикладом вище проведіть аналіз пропущених значень в колонках

Для швидшого аналізу рекомендую написати фукнцію, яку Ви зможете викликати для кожної з колонок.

Зробіть висновок, що робити з пропущеними значеннями в кожному випадку і виконанайте ту дію, яку зазначили.

```python
def basic_eda_function_numeric(dataset, column):
    series = dataset[column]

    result = {
        "head": series.head(),
        "mean": series.mean(),
        "median": series.median(),
        "mode": series.mode()[0] if not series.mode().empty else None,
        "quantiles": series.quantile([0.25, 0.5, 0.75, 1]),
        "missing_pct": series.isna().mean() * 100
    }

    return result
```

### Створення функції `age_cat` для категоризації `YEARS_BORNING`

```python
def age_cat(years):
    if years <= 20:
        return '0-20'
    elif years > 20 and years <= 30:
        return '20-30'
    elif years > 30 and years <= 40:
        return '30-40'
    elif years > 40 and years <= 50:
        return '40-50'
    elif years > 50 and years <= 60:
        return '50-60'
    elif years > 60 and years <= 70:
        return '60-70'
    elif years > 70:
        return '70+'
```

### Одномірний (категоріальний vs цільовий) та двомірний (категоріальний vs категоріальний) аналіз (гістограма) - категоріальні стовпчики

```python
import matplotlib.pyplot as plt


def bi_cat_countplot(df, column, hue_column):
    unique_hue_values = df[hue_column].unique()
    fig, axes = plt.subplots(nrows=1, ncols=2)
    fig.set_size_inches(14,6)

    pltname = f'Нормалізований розподіл значень за категорією: {column}'
    proportions = df.groupby(hue_column)[column].value_counts(normalize=True)
    proportions = (proportions*100).round(2)
    ax = proportions.unstack(hue_column).sort_values(
        by=unique_hue_values[0], ascending=False
        ).plot.bar(ax=axes[0], title=pltname)

    # анотація значень в барплоті
    for container in ax.containers:
        ax.bar_label(container, fmt='{:,.1f}%')

    pltname = f'Кількість даних за категорією: {column}'
    counts = df.groupby(hue_column)[column].value_counts()
    ax = counts.unstack(hue_column).sort_values(
        by=unique_hue_values[0], ascending=False
        ).plot.bar(ax=axes[1], title=pltname)

    for container in ax.containers:
      ax.bar_label(container)


def uni_cat_target_compare(df, column):
    bi_cat_countplot(df, column, hue_column='TARGET' )


def bi_countplot_target(df0, df1, column, hue_column):
  pltname = 'Клієнт зі складнощами щодо платності'
  print(pltname.upper())
  bi_cat_countplot(df1, column, hue_column)
  plt.show()

  pltname = 'Клієнти зі своєчасними платежами'
  print(pltname.upper())
  bi_cat_countplot(df0, column, hue_column)
  plt.show()
```

### Обчислення мінімального та максимального діапазону відхилень для числових стовпців

```python
def outlier_range(dataset,column):
    Q1 = dataset[column].quantile(0.25)
    Q3 = dataset[column].quantile(0.75)
    IQR = Q3 - Q1
    Min_value = (Q1 - 1.5 * IQR)
    Max_value = (Q3 + 1.5 * IQR)
    return Max_value
```

#### Пошук викидів в `AMT_CREDIT` при випадках із вчасними оплатами

Розрахунок IQR (Inter Quartile range)

```python
Q1 = df0['AMT_CREDIT'].quantile(0.25)
Q3 = df0['AMT_CREDIT'].quantile(0.75)
IQR = Q3 - Q1
print(IQR)
```

Значення *до* (Q1 - 1.5 * IQR) та *після* (Q3 + 1.5 * IQR) є викидами.

```python
Min_value0 = (Q1 - 1.5 * IQR)
Max_value0 = (Q3 + 1.5 * IQR)
print("Мінімальне значення, до якого існують викиди: {}".format(Min_value0))
print("Максимальне значення, після якого існують викиди: {}".format(Max_value0))
```

Видалення викидів і побудова діаграми розподілу

```python
def kde_no_outliers(df0, df1, Max_value0, Max_value1, column):
  plt.figure(figsize = (14,6))
  sns.kdeplot(df1[df1[column] <= Max_value1][column],label = 'Payment difficulties')
  sns.kdeplot(df0[df0[column] <= Max_value0][column],label = 'On-Time Payments')
  plt.ticklabel_format(style='plain', axis='x')
  plt.xticks(rotation = 45)
  plt.legend()
  plt.show()
```

```python
kde_no_outliers(df0, df1, Max_value0, Max_value1, 'AMT_CREDIT')
```

**Спостереження**

- Для `AMT_CREDIT` від 250000 до приблизно 650000 більше клієнтів мають труднощі з оплатою
- Для `AMT_CREDIT` > 750000 більше клієнтів, які вчасно здійснюють платежі

[Для цього винесіть функціонал для аналізу у функцію та викличіть функцію для кожної двійки змінних.](https://)
Зробіть висновок про наявність кореляції між змінними.

```python
def analyze_two_columns(df0, df1, col_x, col_y, outlier_func):
    """
    df0 — група з вчасними оплатами
    df1 — група з прострочками
    col_x, col_y — назви колонок для порівняння
    outlier_func — твоя функція outlier_range()
    """

    # Межі викидів
    max0_x = outlier_func(df0, col_x)
    max0_y = outlier_func(df0, col_y)

    max1_x = outlier_func(df1, col_x)
    max1_y = outlier_func(df1, col_y)

    # Фільтрація
    df0_clean = df0[(df0[col_x] < max0_x) & (df0[col_y] < max0_y)]
    df1_clean = df1[(df1[col_x] < max1_x) & (df1[col_y] < max1_y)]

    # Побудова графіків
    plt.figure(figsize=(14, 6))

    # Ліва панель — проблемні платежі
    plt.subplot(1, 2, 1)
    plt.title('Payment difficulties')
    sns.scatterplot(
        x=df1_clean[col_x],
        y=df1_clean[col_y],
        data=df1_clean
    )
    plt.ticklabel_format(style='plain', axis='x')
    plt.ticklabel_format(style='plain', axis='y')

    # Права панель — вчасні платежі
    plt.subplot(1, 2, 2)
    plt.title('On-Time Payments')
    sns.scatterplot(
        x=df0_clean[col_x],
        y=df0_clean[col_y],
        data=df0_clean
    )
    plt.ticklabel_format(style='plain', axis='x')
    plt.ticklabel_format(style='plain', axis='y')

    plt.tight_layout(pad=4)
    plt.show()

    return {
        "max0_x": max0_x,
        "max0_y": max0_y,
        "max1_x": max1_x,
        "max1_y": max1_y
    }
```

```python
analyze_two_columns(df0, df1, 'AMT_ANNUITY', 'AMT_CREDIT', outlier_range)
```

**Спостереження**
- AMT_ANNUITY та AMT_CREDIT демонструють сильну позитивну кореляцію. Це означає, що зі збільшенням суми кредиту зростає і розмір ануїтетного платежу, що є логічним наслідком формули розрахунку ануїтету.

```python
def draw_boxplot(df, categorical, continuous, max_continuous, title, hue_column, subplot_position):
    """
    Малює блок-діаграму для заданого DataFrame, категоріальної та неперервної змінної.
    """
    plt.subplot(1, 2, subplot_position)
    plt.title(title)
    red_diamond = dict(markerfacecolor='r', marker='D')
    sns.boxplot(x=categorical,
                y=df[df[continuous] < max_continuous][continuous],
                data=df,
                flierprops=red_diamond,
                order=sorted(df[categorical].unique(), reverse=True),
                hue=hue_column, hue_order=sorted(df[hue_column].unique(), reverse=True))
    plt.ticklabel_format(style='plain', axis='y')
    plt.xticks(rotation=90)

def bi_boxplot(categorical, continuous, max_continuous1, max_continuous0, hue_column):
    """
    Створює паралельні блок-діаграми для двох груп, визначених у наборі даних, на основі
    категоріальної та неперервної змінної, виділяючи відмінності за допомогою відтінків.
    """
    plt.figure(figsize=(16, 10))

    # Графік для першо групи "Труднощі з платежами" (Payment Difficulties)
    draw_boxplot(df1, categorical, continuous, max_continuous1, 'Payment Difficulties', hue_column, 1)

    # Графік для другої групи "Вчасні оплати" (On-Time Payments)
    draw_boxplot(df0, categorical, continuous, max_continuous0, 'On-Time Payments', hue_column, 2)

    plt.tight_layout(pad=4)
    plt.show()
```

# Висновок: Категорії клієнтів, на яких слід орієнтуватися при наданні кредиту


Після первинного аналізу даних стало зрозуміло, що деякі залежності не є очевидними на загальному рівні, і для формування коректних висновків потрібне додаткове дослідження та попередня фільтрація даних. На цьому етапі можна виділити кілька напрямів, які потребують глибшого аналізу.
Ключові інсайти та напрями для подальшого дослідження
1. Дисбаланс за статтю
- У вибірці спостерігається значний дисбаланс між чоловіками та жінками.
- Через це залежності між статтю та іншими змінними складно інтерпретувати, якщо мета — оцінити реальний вплив статі на ризик.
- Можливо, знадобиться стратифікація або нормалізація за статтю.
2. Аномально низький ризик у групи Businessman
- Відсутність ризикових клієнтів у категорії Businessman виглядає нетипово.
- Потрібно перевірити:
- частку цієї групи у вибірці,
- розмір кредитів, які вони отримують,
- співвідношення кредиту до доходу,
- чи не є ця група занадто малою для статистичних висновків.
- Ймовірно, ця категорія потребує окремого аналізу разом із клієнтами з високими доходами.
3. Вплив витрат для клієнтів із нижчим доходом
- Варто окремо дослідити клієнтів із доходом нижче середнього:
- наявність автомобіля,
- кількість дітей,
- інші потенційні фінансові навантаження.
- Для цієї групи витрати можуть мати значно більший вплив на ризик, ніж у клієнтів із вищими доходами.
4. Освіта як потенційно важливий фактор
- На цьому етапі видно, що вища освіта може позитивно впливати на здатність повернути кредит.
- Це варто перевірити статистично та в моделі.
5. Житлові умови та сімейний стан
- Тип житла (co-op, municipal, living with parents) та статус (single, civil marriage) можуть бути пов’язані з підвищеним ризиком.
- Можливо, це відображає рівень фінансової стабільності або зрілості, але це лише гіпотеза, яку потрібно перевірити.
6. Відсутність інформації про кредитну історію
- Для реального проєкту критично важливо мати дані про:
- попередні кредити,
- історію погашення,
- чи є це перший кредит клієнта.
